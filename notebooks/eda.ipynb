```python
import pandas as pd
```

```python
# Load training data
df = pd.read_csv('../data/train.csv')
data_shape = df.shape
print(f"Loaded Dataframe from ../data/train.csv")
print(f"Dataframe Size: {data_shape[0]} x {data_shape[1]}")

assert df['gpi_value'].notnull().all()
```

```python
# check which columns are null
columns_w_null = df.columns[df.isnull().any()].to_list()
columns_w_null
```

```python
# Find which features have highest proportion of null values
sorted_null_proportions = (df.isnull()
    .mean()
    .rename('proportion_null')
    .to_frame()
    .sort_values(by="proportion_null", ascending=False)
)
sorted_null_proportions
```
